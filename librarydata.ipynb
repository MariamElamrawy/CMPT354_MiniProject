# Creating tables

Creating item table

```python
%%sql
CREATE TABLE Item (
    item_id       INTEGER PRIMARY KEY,
    title         TEXT NOT NULL,
    creator       TEXT,
    publisher     TEXT,
    year          INTEGER,
    subject_genre TEXT,
    language      TEXT
);
```

Creating table for PrintBook

```python
%%sql
CREATE TABLE PrintBook (
    item_id   INTEGER PRIMARY KEY,
    ISBN      TEXT UNIQUE,
    page_num  INTEGER,
    FOREIGN KEY (item_id) REFERENCES Item(item_id)
);
```

Creating EBook table

```python
%%sql
CREATE TABLE EBook (
    item_id   INTEGER PRIMARY KEY,
    ISBN      TEXT UNIQUE,
    format    TEXT,
    FOREIGN KEY (item_id) REFERENCES Item(item_id)
);
```

Creating Magazine table

```python
%%sql
CREATE TABLE Magazine (
    item_id     INTEGER PRIMARY KEY,
    ISSN        TEXT,
    issue       TEXT,
    issue_date  DATE,
    FOREIGN KEY (item_id) REFERENCES Item(item_id)
);
```

Creating Journal table

```python
%%sql
CREATE TABLE Journal (
    item_id   INTEGER PRIMARY KEY,
    ISSN      TEXT,
    issue     TEXT,
    volume    INTEGER,
    FOREIGN KEY (item_id) REFERENCES Item(item_id)
);
```

Creating record table

```python
%%sql
CREATE TABLE Record (
    item_id   INTEGER PRIMARY KEY,
    type      TEXT,
    quantity  INTEGER,
    FOREIGN KEY (item_id) REFERENCES Item(item_id)
);
```

Creating Copy Table

```python
%%sql
CREATE TABLE Copy (
    item_id      INTEGER,
    copy_number  INTEGER,
    status       TEXT,
    PRIMARY KEY (item_id, copy_number),
    FOREIGN KEY (item_id) REFERENCES Item(item_id)
);
```

Creating Member Table

```python
%%sql
CREATE TABLE Member (
    member_id  INTEGER PRIMARY KEY,
    name       TEXT NOT NULL,
    address    TEXT,
    phone      TEXT,
    email      TEXT,
    reg_date   DATE,
    status     TEXT
);
```

Creating Loan Table

```python
%%sql
CREATE TABLE Loan (
    loan_id        INTEGER PRIMARY KEY,
    checkout_date  DATE,
    due_date       DATE,
    returned_date  DATE,
    item_id        INTEGER,
    copy_number    INTEGER,
    member_id      INTEGER,
    FOREIGN KEY (item_id, copy_number) REFERENCES Copy(item_id, copy_number),
    FOREIGN KEY (member_id) REFERENCES Member(member_id)
);
```

Creating Fine

```python
%%sql
CREATE TABLE Fine (
    loan_id      INTEGER PRIMARY KEY,
    date_issued  DATE,
    amount       REAL,
    status       TEXT,
    FOREIGN KEY (loan_id) REFERENCES Loan(loan_id)
);
```

Creating Room table

```python
%%sql
CREATE TABLE Room (
    room_id   INTEGER PRIMARY KEY,
    name      TEXT,
    capacity  INTEGER
);
```

Creating event table

```python
%%sql
CREATE TABLE Event (
    event_id   INTEGER PRIMARY KEY,
    name       TEXT NOT NULL,
    type       TEXT,
    date       DATE,
    time       TEXT,
    audience   TEXT,
    capacity   INTEGER,
    room_id    INTEGER,
    FOREIGN KEY (room_id) REFERENCES Room(room_id)
);
```

Creating personnel Table

```python
%%sql
CREATE TABLE Personnel (
    employee_id  INTEGER PRIMARY KEY,
    first_name   TEXT NOT NULL,
    last_name    TEXT NOT NULL,
    role         TEXT,
    email        TEXT,
    phone        TEXT
);
```

Creating Hold relation

A specific item on hold for a member.

```python
%%sql
CREATE TABLE Hold (
    member_id    INTEGER,
    item_id      INTEGER,
    date_placed  DATE,
    PRIMARY KEY (member_id, item_id),
    FOREIGN KEY (member_id) REFERENCES Member(member_id),
    FOREIGN KEY (item_id) REFERENCES Item(item_id)
);
```

Creating "Registers" Relation

A member regestiering for a specific event.

```python
%%sql
CREATE TABLE Registers (
    member_id  INTEGER,
    event_id   INTEGER,
    reg_date   DATE,
    PRIMARY KEY (member_id, event_id),
    FOREIGN KEY (member_id) REFERENCES Member(member_id),
    FOREIGN KEY (event_id) REFERENCES Event(event_id)
);
```

Creating "Volunteers" Relations

Where the volunteers are a member of the library and volunteer for specific event.

```python
%%sql
CREATE TABLE Volunteers (
    member_id  INTEGER,
    event_id   INTEGER,
    PRIMARY KEY (member_id, event_id),
    FOREIGN KEY (member_id) REFERENCES Member(member_id),
    FOREIGN KEY (event_id) REFERENCES Event(event_id)
);
```

Creating "Manages" Relation

Where a personnel manages a specific item.

```python
%%sql
CREATE TABLE Manages (
    employee_id  INTEGER,
    item_id      INTEGER,
    PRIMARY KEY (employee_id, item_id),
    FOREIGN KEY (employee_id) REFERENCES Personnel(employee_id),
    FOREIGN KEY (item_id) REFERENCES Item(item_id)
);
```

# Constraints

```python
%%sql
DROP TABLE Copy;

CREATE TABLE Copy (
    item_id      INTEGER,
    copy_number  INTEGER,
    status       TEXT CHECK (status IN ('available', 'on loan', 'on hold', 'lost', 'in repair', 'reference-only')),
    PRIMARY KEY (item_id, copy_number),
    FOREIGN KEY (item_id) REFERENCES Item(item_id)
);
```

```python
%%sql
DROP TABLE Member;

CREATE TABLE Member (
    member_id  INTEGER PRIMARY KEY,
    name       TEXT NOT NULL,
    address    TEXT,
    phone      TEXT,
    email      TEXT,
    reg_date   DATE,
    status     TEXT CHECK (status IN ('active', 'expired', 'suspended'))
);
```

```python
%%sql
DROP TABLE Fine;

CREATE TABLE Fine (
    loan_id      INTEGER PRIMARY KEY,
    date_issued  DATE,
    amount       REAL,
    status       TEXT CHECK (status IN ('unpaid', 'paid')),
    FOREIGN KEY (loan_id) REFERENCES Loan(loan_id)
);
```

Creating "Donates" Relation

One donor per physical copy, so (item_id, copy_number) is the key.

```python
%%sql
CREATE TABLE Donates (
    item_id       INTEGER,
    copy_number   INTEGER,
    member_id     INTEGER,
    date_donated  DATE,
    PRIMARY KEY (item_id, copy_number),
    FOREIGN KEY (item_id, copy_number) REFERENCES Copy(item_id, copy_number),
    FOREIGN KEY (member_id) REFERENCES Member(member_id)
);
```

# Trigger

```python
%%sql
CREATE TRIGGER copy_on_loan
AFTER INSERT ON Loan
BEGIN
    UPDATE Copy
    SET status = 'on loan'
    WHERE item_id = NEW.item_id AND copy_number = NEW.copy_number;
END;
```

```python
%%sql
CREATE TRIGGER copy_returned
AFTER UPDATE OF returned_date ON Loan
WHEN NEW.returned_date IS NOT NULL
BEGIN
    UPDATE Copy
    SET status = 'available'
    WHERE item_id = NEW.item_id AND copy_number = NEW.copy_number;
END;
```

```python
%%sql
CREATE TRIGGER block_loan_if_fines_exceed
BEFORE INSERT ON Loan
WHEN (
    SELECT COALESCE(SUM(f.amount), 0)
    FROM Fine f
    JOIN Loan l ON f.loan_id = l.loan_id
    WHERE l.member_id = NEW.member_id AND f.status = 'unpaid'
) > 20
BEGIN
    SELECT RAISE(ABORT, 'Cannot borrow: unpaid fines exceed threshold');
END;
```

```python
%%sql
CREATE TRIGGER block_hold_if_available
BEFORE INSERT ON Hold
WHEN (
    SELECT COUNT(*)
    FROM Copy
    WHERE item_id = NEW.item_id AND status = 'available'
) > 0
BEGIN
    SELECT RAISE(ABORT, 'Cannot place hold: a copy is currently available');
END;
```
